Load MarianMT models

```python

from transformers import MarianMTModel, MarianTokenizer
import torch
import gc

MODEL_CONFIG = {
    "Hindi": "Helsinki-NLP/opus-mt-en-hi",
    "Marathi": "Helsinki-NLP/opus-mt-en-mr"
}

loaded_models = {}

for language, model_name in MODEL_CONFIG.items():
    print(f"\nLoading {language}: {model_name}")

    tokenizer = MarianTokenizer.from_pretrained(model_name)

    model = MarianMTModel.from_pretrained(model_name)
    model = model.to(DEVICE)
    model.eval()

    loaded_models[language] = {
        "tokenizer": tokenizer,
        "model": model,
        "model_name": model_name
    }

    print(f"{language} model loaded successfully!")

print("\nBoth translation models are ready.")
```

Translation function

```python

import time

def translate_text(text, target_language="Hindi"):
    if not isinstance(text, str) or not text.strip():
        return {"success": False, "error": "Please enter valid text."}

    if target_language not in loaded_models:
        return {"success": False, "error": "Unsupported target language."}

    tokenizer = loaded_models[target_language]["tokenizer"]
    model = loaded_models[target_language]["model"]

    try:
        start_time = time.time()

        inputs = tokenizer(
            text,
            return_tensors="pt",
            truncation=True,
            max_length=512
        )

        inputs = {key: value.to(DEVICE) for key, value in inputs.items()}

        with torch.inference_mode():
            generated = model.generate(
                **inputs,
                max_new_tokens=256,
                num_beams=4
            )

        translated = tokenizer.decode(
            generated[0],
            skip_special_tokens=True
        )

        latency = round(time.time() - start_time, 3)

        return {
            "success": True,
            "original": text,
            "translated": translated,
            "target_language": target_language,
            "model": loaded_models[target_language]["model_name"],
            "device": DEVICE,
            "latency_seconds": latency
        }

    except Exception as error:
        return {
            "success": False,
            "error": str(error)
        }
```

prompt engineering module

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"
MODULE_DIR.mkdir(exist_ok=True)

code = r'''
"""
LinguaBridge AI - Prompt Engineering Module

Supports zero-shot, few-shot and role-based prompts,
along with different translation tones and registers.

These prompts are designed for instruction-following LLMs.
MarianMT is evaluated separately as a translation baseline.
"""

SUPPORTED_LANGUAGES = {
    "Hindi": "Hindi",
    "Marathi": "Marathi"
}

SUPPORTED_STYLES = {
    "Neutral": "Translate naturally and accurately.",
    "Formal": "Use formal, respectful and grammatically correct language.",
    "Informal": "Use casual and friendly everyday language.",
    "Conversational": "Use natural spoken language as in everyday conversation.",
    "Academic": "Use precise academic vocabulary and a scholarly register.",
    "Professional": "Use polished professional language suitable for workplace communication."
}

SUPPORTED_TECHNIQUES = [
    "zero_shot",
    "few_shot",
    "role_based"
]


def validate_inputs(text, target_language, style="Neutral"):
    """Validate translator inputs."""

    if not isinstance(text, str) or not text.strip():
        raise ValueError("Input text cannot be empty.")

    if target_language not in SUPPORTED_LANGUAGES:
        raise ValueError("Target language must be Hindi or Marathi.")

    if style not in SUPPORTED_STYLES:
        raise ValueError(
            f"Unsupported style. Choose from: {list(SUPPORTED_STYLES)}"
        )

    return True


def build_zero_shot_prompt(text, target_language, style="Neutral"):
    """Create a direct translation instruction."""

    validate_inputs(text, target_language, style)

    return f"""
Translate the following English text into {target_language}.

Requirements:
- Preserve the original meaning.
- Do not omit important information.
- Do not add explanations.
- Maintain grammatical correctness.
- {SUPPORTED_STYLES[style]}

English text:
{text}

Return only the translated text.
""".strip()


def build_few_shot_prompt(text, target_language, style="Neutral", examples=None):
    """Create a prompt containing translation examples."""

    validate_inputs(text, target_language, style)

    if examples is None:
        examples = {
            "Hindi": [
                ("Good morning.", "सुप्रभात।"),
                ("How are you?", "आप कैसे हैं?")
            ],
            "Marathi": [
                ("Good morning.", "शुभ सकाळ."),
                ("How are you?", "तुम्ही कसे आहात?")
            ]
        }

    prompt = f"""
Translate English into {target_language}.

Follow the examples below for translation quality and style.

Style: {SUPPORTED_STYLES[style]}

"""

    for source, translation in examples.get(target_language, []):
        prompt += f"English: {source}\n{target_language}: {translation}\n\n"

    prompt += f"""
Now translate this sentence:

English: {text}
{target_language}:
"""

    return prompt.strip()


def build_role_based_prompt(text, target_language, style="Neutral"):
    """Create a role-based translation prompt."""

    validate_inputs(text, target_language, style)

    return f"""
You are an expert English-to-{target_language} translator.

Your responsibilities:
1. Produce an accurate translation.
2. Preserve meaning and context.
3. Use culturally appropriate expressions.
4. Avoid unnecessary additions or omissions.
5. Follow the requested language register.

Required style:
{SUPPORTED_STYLES[style]}

Text to translate:
{text}

Return only the {target_language} translation.
""".strip()


def build_translation_prompt(
    text,
    target_language="Hindi",
    style="Neutral",
    technique="zero_shot"
):
    """Main prompt builder."""

    builders = {
        "zero_shot": build_zero_shot_prompt,
        "few_shot": build_few_shot_prompt,
        "role_based": build_role_based_prompt
    }

    if technique not in builders:
        raise ValueError(
            f"Unsupported technique. Choose from: {SUPPORTED_TECHNIQUES}"
        )

    return builders[technique](text, target_language, style)


def compare_prompt_techniques(text, target_language="Hindi", style="Formal"):
    """Generate prompts for comparing different techniques."""

    return {
        technique: build_translation_prompt(
            text=text,
            target_language=target_language,
            style=style,
            technique=technique
        )
        for technique in SUPPORTED_TECHNIQUES
    }
'''

module_path = MODULE_DIR / "prompt_engineering.py"
module_path.write_text(code.strip() + "\n", encoding="utf-8")

print("Prompt engineering module created:")
print(module_path)
```

Create the LLM translator module

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"
MODULE_DIR.mkdir(parents=True, exist_ok=True)

code = r'''
"""
LinguaBridge AI - Qwen LLM Translation Engine

Uses a locally loaded Qwen2.5 instruction-tuned model.
Supports Hindi and Marathi translation with configurable
prompt-engineering techniques.
"""

import time
import torch

from transformers import AutoTokenizer, AutoModelForCausalLM

from modules.prompt_engineering import build_translation_prompt


MODEL_NAME = "Qwen/Qwen2.5-1.5B-Instruct"

SUPPORTED_LANGUAGES = ["Hindi", "Marathi"]

_tokenizer = None
_model = None
_device = None


def load_llm():
    """
    Load Qwen once and reuse it for subsequent translations.
    Uses FP16 on CUDA and FP32 on CPU.
    """

    global _tokenizer, _model, _device

    if _model is not None:
        return True

    _device = "cuda" if torch.cuda.is_available() else "cpu"

    dtype = torch.float16 if _device == "cuda" else torch.float32

    print(f"Loading {MODEL_NAME} on {_device}...")

    _tokenizer = AutoTokenizer.from_pretrained(
        MODEL_NAME,
        trust_remote_code=False
    )

    _model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        torch_dtype=dtype,
        trust_remote_code=False
    )

    _model.to(_device)
    _model.eval()

    print("Qwen model loaded successfully.")

    return True


def translate_with_llm(
    text,
    target_language="Hindi",
    style="Neutral",
    technique="zero_shot",
    max_new_tokens=160
):
    """
    Translate English text into Hindi or Marathi.

    Supported techniques:
    - zero_shot
    - few_shot
    - role_based
    """

    if not isinstance(text, str) or not text.strip():
        return {
            "success": False,
            "error": "Input text cannot be empty."
        }

    if target_language not in SUPPORTED_LANGUAGES:
        return {
            "success": False,
            "error": "Unsupported target language."
        }

    try:
        # Build prompt using Module 2
        prompt = build_translation_prompt(
            text=text,
            target_language=target_language,
            style=style,
            technique=technique
        )

        # Load model if not already loaded
        load_llm()

        messages = [
            {
                "role": "system",
                "content": (
                    "You are a professional English-to-Indian-language "
                    "translator. Follow the translation instructions "
                    "carefully. Return only the translated text."
                )
            },
            {
                "role": "user",
                "content": prompt
            }
        ]

        chat_text = _tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )

        inputs = _tokenizer(
            chat_text,
            return_tensors="pt",
            truncation=True,
            max_length=2048
        )

        inputs = {
            key: value.to(_device)
            for key, value in inputs.items()
        }

        start_time = time.time()

        with torch.inference_mode():
            outputs = _model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
                repetition_penalty=1.05,
                pad_token_id=_tokenizer.eos_token_id,
                eos_token_id=_tokenizer.eos_token_id
            )

        # Decode only newly generated tokens
        input_length = inputs["input_ids"].shape[1]

        generated_tokens = outputs[0][input_length:]

        translated_text = _tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        ).strip()

        latency = round(time.time() - start_time, 3)

        return {
            "success": True,
            "original": text,
            "translated": translated_text,
            "target_language": target_language,
            "style": style,
            "technique": technique,
            "model": MODEL_NAME,
            "device": _device,
            "latency_seconds": latency
        }

    except Exception as error:
        return {
            "success": False,
            "error": str(error)
        }


def unload_llm():
    """Release model memory when necessary."""

    global _tokenizer, _model, _device

    if _model is not None:
        del _model
        del _tokenizer

    _model = None
    _tokenizer = None
    _device = None

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    print("LLM unloaded and memory released.")
'''

module_path = MODULE_DIR / "llm_translator.py"
module_path.write_text(code.strip() + "\n", encoding="utf-8")

print("Created:", module_path)
```

Create the evaluation engine

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"

evaluation_code = r'''
"""LinguaBridge AI - Translation evaluation engine."""

import time
import pandas as pd
import sacrebleu


def calculate_metrics(predictions, references):
    """
    Calculate corpus BLEU and chrF.
    Predictions and references must be aligned lists.
    """

    if len(predictions) != len(references):
        raise ValueError("Predictions and references must have equal lengths.")

    if not predictions:
        raise ValueError("No translations provided.")

    if not all(str(ref).strip() for ref in references):
        raise ValueError("All reference translations must be populated.")

    bleu = sacrebleu.corpus_bleu(
        predictions,
        [references]
    )

    chrf = sacrebleu.corpus_chrf(
        predictions,
        [references]
    )

    return {
        "BLEU": round(bleu.score, 3),
        "chrF": round(chrf.score, 3)
    }


def run_model_comparison(
    sentences,
    translation_functions,
    target_language
):
    """
    Run each model against the same set of sentences.

    translation_functions:
        Dictionary mapping model names to callable functions.
    """

    results = []

    for sentence_id, sentence in enumerate(sentences, start=1):

        for model_name, translate_function in translation_functions.items():

            start_time = time.time()

            try:
                output = translate_function(
                    sentence,
                    target_language
                )

                elapsed = round(time.time() - start_time, 3)

                if isinstance(output, dict):
                    success = output.get("success", False)
                    translated = output.get("translated", "")
                    error = output.get("error", "")
                else:
                    success = bool(output)
                    translated = str(output) if output else ""
                    error = ""

                results.append({
                    "id": sentence_id,
                    "english": sentence,
                    "target_language": target_language,
                    "model": model_name,
                    "translation": translated,
                    "latency_seconds": elapsed,
                    "success": success,
                    "error": error
                })

            except Exception as exc:
                results.append({
                    "id": sentence_id,
                    "english": sentence,
                    "target_language": target_language,
                    "model": model_name,
                    "translation": "",
                    "latency_seconds": None,
                    "success": False,
                    "error": str(exc)
                })

    return pd.DataFrame(results)


def summarize_latency(results):
    """Summarize successful model inference latency."""

    successful = results[results["success"] == True]

    if successful.empty:
        return pd.DataFrame()

    return (
        successful.groupby("model")["latency_seconds"]
        .agg(["mean", "median", "min", "max", "count"])
        .round(3)
        .reset_index()
    )


def evaluate_against_references(
    results,
    reference_column,
    benchmark_df
):
    """Calculate metrics separately for each model."""

    merged = results.merge(
        benchmark_df[["id", reference_column]],
        on="id",
        how="left"
    )

    summaries = []

    for model_name, group in merged.groupby("model"):

        valid = group[
            (group["success"] == True) &
            group[reference_column].notna() &
            (group[reference_column].astype(str).str.strip() != "")
        ]

        if valid.empty:
            summaries.append({
                "model": model_name,
                "BLEU": None,
                "chrF": None,
                "evaluated_sentences": 0,
                "status": "Reference translations missing"
            })
            continue

        scores = calculate_metrics(
            valid["translation"].astype(str).tolist(),
            valid[reference_column].astype(str).tolist()
        )

        summaries.append({
            "model": model_name,
            **scores,
            "evaluated_sentences": len(valid),
            "status": "Calculated"
        })

    return pd.DataFrame(summaries)
'''

(MODULE_DIR / "evaluation.py").write_text(
    evaluation_code.strip() + "\n",
    encoding="utf-8"
)

print("Evaluation module created.")
```

Connect the two translation models

```python

import torch
import time
from transformers import MarianMTModel, MarianTokenizer

from modules.llm_translator import translate_with_llm

MARIAN_MODELS = {
    "Hindi": "Helsinki-NLP/opus-mt-en-hi",
    "Marathi": "Helsinki-NLP/opus-mt-en-mr"
}

marian_cache = {}

def load_marian_for_evaluation(language):
    if language not in MARIAN_MODELS:
        raise ValueError("Unsupported language")

    if language not in marian_cache:
        name = MARIAN_MODELS[language]
        print("Loading MarianMT:", name)

        tokenizer = MarianTokenizer.from_pretrained(name)
        model = MarianMTModel.from_pretrained(name)

        device = "cuda" if torch.cuda.is_available() else "cpu"
        model.to(device)
        model.eval()

        marian_cache[language] = {
            "tokenizer": tokenizer,
            "model": model,
            "device": device
        }

    return marian_cache[language]


def marian_translate_for_evaluation(text, language):
    if not isinstance(text, str) or not text.strip():
        return {"success": False, "error": "Empty input"}

    try:
        bundle = load_marian_for_evaluation(language)

        inputs = bundle["tokenizer"](
            text,
            return_tensors="pt",
            truncation=True,
            max_length=512
        )

        inputs = {
            key: value.to(bundle["device"])
            for key, value in inputs.items()
        }

        start = time.time()

        with torch.inference_mode():
            output = bundle["model"].generate(
                **inputs,
                max_new_tokens=256,
                num_beams=4
            )

        translation = bundle["tokenizer"].decode(
            output[0],
            skip_special_tokens=True
        )

        return {
            "success": True,
            "translated": translation,
            "latency_seconds": round(time.time() - start, 3)
        }

    except Exception as e:
        return {"success": False, "error": str(e)}


def qwen_translate_for_evaluation(text, language):
    return translate_with_llm(
        text=text,
        target_language=language,
        style="Neutral",
        technique="zero_shot",
        max_new_tokens=160
    )


translation_functions = {
    "MarianMT": marian_translate_for_evaluation,
    "Qwen2.5-1.5B": qwen_translate_for_evaluation
}

print("Model adapters ready.")
print("MarianMT and Qwen comparison functions registered.")
```

Create the BLIP image captioning module

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"
MODULE_DIR.mkdir(parents=True, exist_ok=True)

code = r'''
"""
LinguaBridge AI - BLIP Image Captioning Module

Generates English image captions using a locally loaded
Salesforce BLIP image-captioning model.
"""

import torch
from PIL import Image, UnidentifiedImageError
from transformers import BlipProcessor, BlipForConditionalGeneration


MODEL_NAME = "Salesforce/blip-image-captioning-base"

_processor = None
_model = None
_device = None


def load_blip():
    """Load BLIP once and reuse the model."""

    global _processor, _model, _device

    if _model is not None:
        return True

    _device = "cuda" if torch.cuda.is_available() else "cpu"

    print(f"Loading BLIP model on {_device}...")

    _processor = BlipProcessor.from_pretrained(MODEL_NAME)

    dtype = torch.float16 if _device == "cuda" else torch.float32

    _model = BlipForConditionalGeneration.from_pretrained(
        MODEL_NAME,
        torch_dtype=dtype
    )

    _model.to(_device)
    _model.eval()

    print("BLIP model loaded successfully.")

    return True


def caption_image(image_path, max_new_tokens=60):
    """
    Generate an English caption for an image.

    Returns a structured dictionary containing the caption,
    model name, device and inference latency.
    """

    if not image_path:
        return {
            "success": False,
            "error": "Please provide an image path."
        }

    try:
        image = Image.open(image_path).convert("RGB")

    except (FileNotFoundError, UnidentifiedImageError, OSError) as exc:
        return {
            "success": False,
            "error": f"Unable to open image: {str(exc)}"
        }

    try:
        load_blip()

        inputs = _processor(
            images=image,
            return_tensors="pt"
        )

        inputs = {
            key: value.to(_device)
            for key, value in inputs.items()
        }

        if _device == "cuda":
            inputs["pixel_values"] = inputs["pixel_values"].half()

        import time
        start = time.time()

        with torch.inference_mode():
            output = _model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                num_beams=4
            )

        caption = _processor.decode(
            output[0],
            skip_special_tokens=True
        ).strip()

        latency = round(time.time() - start, 3)

        return {
            "success": True,
            "caption": caption,
            "model": MODEL_NAME,
            "device": _device,
            "latency_seconds": latency,
            "image_size": image.size
        }

    except Exception as exc:
        return {
            "success": False,
            "error": str(exc)
        }


def unload_blip():
    """Release BLIP model memory."""

    global _processor, _model, _device

    _model = None
    _processor = None
    _device = None

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    print("BLIP model unloaded.")
'''

module_path = MODULE_DIR / "image_caption.py"
module_path.write_text(code.strip() + "\n", encoding="utf-8")

print("BLIP module created:", module_path)
```

Create the reusable image translation pipeline

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"

pipeline_code = r'''
"""
LinguaBridge AI - Image-to-Translation Pipeline

Pipeline:
Image -> BLIP caption -> MarianMT translation
"""

from modules.image_caption import caption_image


def process_image(
    image_path,
    target_language="Hindi",
    translation_function=None
):
    """
    Generate an English caption and translate it.

    translation_function must accept:
        translation_function(text, target_language)

    Returns caption, translation and model metadata.
    """

    if translation_function is None:
        raise ValueError(
            "A translation_function must be provided."
        )

    caption_result = caption_image(image_path)

    if not caption_result.get("success"):
        return {
            "success": False,
            "stage": "captioning",
            "error": caption_result.get("error")
        }

    caption = caption_result["caption"]

    translation_result = translation_function(
        caption,
        target_language
    )

    if not translation_result.get("success"):
        return {
            "success": False,
            "stage": "translation",
            "caption": caption,
            "error": translation_result.get("error")
        }

    return {
        "success": True,
        "image_path": str(image_path),
        "english_caption": caption,
        "translated_caption": translation_result["translated"],
        "target_language": target_language,
        "caption_model": caption_result["model"],
        "translation_model": translation_result.get("model", "Unknown"),
        "caption_latency_seconds": caption_result["latency_seconds"],
        "translation_latency_seconds": translation_result.get(
            "latency_seconds"
        )
    }
'''

pipeline_path = MODULE_DIR / "image_translation.py"
pipeline_path.write_text(pipeline_code.strip() + "\n", encoding="utf-8")

print("Image translation pipeline created:", pipeline_path)
```

```python

from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"

speech_code = r'''
"""
LinguaBridge AI - Indic Parler Text-to-Speech Module

Converts Hindi and Marathi text into speech using
AI4Bharat Indic Parler-TTS.
"""

import time
import torch
import soundfile as sf
import numpy as np

from pathlib import Path
from transformers import AutoTokenizer
from parler_tts import ParlerTTSForConditionalGeneration


MODEL_NAME = "ai4bharat/indic-parler-tts"

SUPPORTED_SPEAKERS = {
    "Hindi": ["Divya", "Rohit", "Aman", "Rani"],
    "Marathi": ["Sunita", "Sanjay", "Nikhil", "Radha", "Varun", "Isha"]
}

SPEAKER_DEFAULTS = {
    "Hindi": "Divya",
    "Marathi": "Sunita"
}

VOICE_STYLES = {
    "Normal": "speaks at a moderate speed and pitch in a clear, neutral tone",
    "Slow": "speaks slowly with clear pronunciation and a calm tone",
    "Expressive": "speaks with an expressive, engaging and natural tone"
}

_model = None
_tokenizer = None
_description_tokenizer = None
_device = None


def load_tts():
    """Load the TTS model and tokenizers once."""

    global _model, _tokenizer, _description_tokenizer, _device

    if _model is not None:
        return True

    _device = "cuda" if torch.cuda.is_available() else "cpu"

    dtype = torch.float16 if _device == "cuda" else torch.float32

    print(f"Loading Indic Parler-TTS on {_device}...")

    _model = ParlerTTSForConditionalGeneration.from_pretrained(
        MODEL_NAME,
        torch_dtype=dtype
    ).to(_device)

    _model.eval()

    _tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

    _description_tokenizer = AutoTokenizer.from_pretrained(
        _model.config.text_encoder._name_or_path
    )

    print("Indic Parler-TTS loaded successfully.")

    return True


def generate_speech(
    text,
    language="Hindi",
    speaker=None,
    style="Normal",
    output_path=None
):
    """
    Convert Hindi or Marathi text to a WAV audio file.

    Returns structured output with file path and metadata.
    """

    if not isinstance(text, str) or not text.strip():
        return {
            "success": False,
            "error": "Text cannot be empty."
        }

    if language not in SUPPORTED_SPEAKERS:
        return {
            "success": False,
            "error": "Only Hindi and Marathi are supported."
        }

    if style not in VOICE_STYLES:
        return {
            "success": False,
            "error": "Invalid speech style."
        }

    speaker = speaker or SPEAKER_DEFAULTS[language]

    if speaker not in SUPPORTED_SPEAKERS[language]:
        return {
            "success": False,
            "error": f"Unsupported {language} speaker: {speaker}"
        }

    try:
        load_tts()

        description = (
            f"{speaker} {VOICE_STYLES[style]}. "
            "The recording is of very high quality, "
            "with very clear audio and no background noise."
        )

        prompt_inputs = _tokenizer(
            text.strip(),
            return_tensors="pt"
        ).to(_device)

        description_inputs = _description_tokenizer(
            description,
            return_tensors="pt"
        ).to(_device)

        start_time = time.time()

        with torch.inference_mode():
            generation = _model.generate(
                input_ids=description_inputs.input_ids,
                attention_mask=description_inputs.attention_mask,
                prompt_input_ids=prompt_inputs.input_ids,
                prompt_attention_mask=prompt_inputs.attention_mask
            )

        audio = generation.detach().float().cpu().numpy().squeeze()

        if audio.ndim != 1 or audio.size == 0:
            raise ValueError("Model returned invalid audio output.")

        PROJECT_DIR = Path("/content/LinguaBridge-AI")
        output_dir = PROJECT_DIR / "outputs" / "audio"
        output_dir.mkdir(parents=True, exist_ok=True)

        if output_path is None:
            filename = f"{language.lower()}_{int(time.time())}.wav"
            output_path = output_dir / filename
        else:
            output_path = Path(output_path)
            output_path.parent.mkdir(parents=True, exist_ok=True)

        sample_rate = _model.config.sampling_rate

        sf.write(
            str(output_path),
            audio,
            sample_rate
        )

        latency = round(time.time() - start_time, 3)

        return {
            "success": True,
            "audio_path": str(output_path),
            "language": language,
            "speaker": speaker,
            "style": style,
            "model": MODEL_NAME,
            "device": _device,
            "sample_rate": sample_rate,
            "duration_seconds": round(len(audio) / sample_rate, 2),
            "latency_seconds": latency
        }

    except Exception as exc:
        return {
            "success": False,
            "error": str(exc)
        }


def unload_tts():
    """Release TTS model memory."""

    global _model, _tokenizer, _description_tokenizer, _device

    _model = None
    _tokenizer = None
    _description_tokenizer = None
    _device = None

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

    print("TTS model unloaded.")
'''

speech_path = MODULE_DIR / "speech.py"
speech_path.write_text(speech_code.strip() + "\n", encoding="utf-8")

print("TTS module created:", speech_path)
```

```python
from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"

MODULE_DIR.mkdir(parents=True, exist_ok=True)

translation_code = r'''
import torch

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from IndicTransToolkit.processor import IndicProcessor


MODEL_NAME = "ai4bharat/indictrans2-en-indic-dist-200M"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

LANGUAGE_CODES = {
    "Hindi": "hin_Deva",
    "Marathi": "mar_Deva"
}

tokenizer = None
model = None
processor = None


def load_translation_model():

    global tokenizer, model, processor

    if model is not None:
        return True

    try:
        print("Loading IndicTrans2 translation model...")

        tokenizer = AutoTokenizer.from_pretrained(
            MODEL_NAME,
            trust_remote_code=True
        )

        model = AutoModelForSeq2SeqLM.from_pretrained(
            MODEL_NAME,
            trust_remote_code=True,
            low_cpu_mem_usage=True
        )

        model = model.to(DEVICE)
        model.eval()

        processor = IndicProcessor(inference=True)

        print("Translation model loaded successfully!")
        print("Device:", DEVICE)

        return True

    except Exception as e:
        model = None
        tokenizer = None
        processor = None

        raise RuntimeError(
            f"Translation model loading failed: {e}"
        ) from e


def translate_text(text, target_language):

    if not isinstance(text, str) or not text.strip():
        return {
            "success": False,
            "translated": None,
            "error": "Input text cannot be empty."
        }

    if target_language not in LANGUAGE_CODES:
        return {
            "success": False,
            "translated": None,
            "error": f"Unsupported language: {target_language}"
        }

    try:

        if model is None:
            load_translation_model()

        source_language = "eng_Latn"
        target_code = LANGUAGE_CODES[target_language]

        batch = processor.preprocess_batch(
            [text.strip()],
            src_lang=source_language,
            tgt_lang=target_code
        )

        inputs = tokenizer(
            batch,
            truncation=True,
            padding="longest",
            return_tensors="pt",
            return_attention_mask=True
        ).to(DEVICE)

        with torch.inference_mode():

            generated_tokens = model.generate(
                **inputs,
                use_cache=True,
                min_length=1,
                max_length=256,
                num_beams=5,
                num_return_sequences=1
            )

        generated_tokens = tokenizer.batch_decode(
            generated_tokens,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=True
        )

        translations = processor.postprocess_batch(
            generated_tokens,
            lang=target_code
        )

        return {
            "success": True,
            "translated": translations[0],
            "source_language": "English",
            "target_language": target_language,
            "error": None
        }

    except Exception as e:

        return {
            "success": False,
            "translated": None,
            "error": str(e)
        }
'''

translation_file = MODULE_DIR / "translation.py"

translation_file.write_text(
    translation_code,
    encoding="utf-8"
)

print("Translation module created!")
print("Location:", translation_file)
```

Create the gender-bias evaluation module

```python

from pathlib import Path

# ==========================================
# 1. PROJECT DIRECTORY SETUP
# ==========================================

PROJECT_DIR = Path("/content/LinguaBridge-AI")
MODULE_DIR = PROJECT_DIR / "modules"

# Create all missing parent directories
MODULE_DIR.mkdir(parents=True, exist_ok=True)

print("Project directory:", PROJECT_DIR)
print("Modules directory:", MODULE_DIR)


# ==========================================
# 2. BIAS EVALUATION MODULE
# ==========================================

bias_code = r'''
"""
LinguaBridge AI - Gender Bias Evaluation

Screens translations for selected masculine and feminine
grammatical markers.

The marker lists are heuristic, not a complete grammar analyser.
Results require human linguistic review.
"""

import pandas as pd


GENDER_MARKERS = {
    "Hindi": {
        "masculine": [
            "करता", "गया", "रहा", "था", "बैठा",
            "लिखता", "बोला", "आया", "सकता", "हुआ"
        ],
        "feminine": [
            "करती", "गई", "रही", "थी", "बैठी",
            "लिखती", "बोली", "आई", "सकती", "हुई"
        ]
    },

    "Marathi": {
        "masculine": [
            "करतो", "गेला", "होता", "बसला",
            "लिहितो", "बोलला", "आला", "शकतो"
        ],
        "feminine": [
            "करते", "गेली", "होती", "बसली",
            "लिहिते", "बोलली", "आली", "शकते"
        ]
    }
}


def detect_gender_markers(text, language):
    """
    Detect selected grammatical gender markers.

    This is a heuristic screening function.
    """

    if language not in GENDER_MARKERS:
        raise ValueError(
            "Language must be Hindi or Marathi."
        )

    if not isinstance(text, str):
        text = ""

    text = text.strip()

    masculine = [
        marker
        for marker in GENDER_MARKERS[language]["masculine"]
        if marker in text
    ]

    feminine = [
        marker
        for marker in GENDER_MARKERS[language]["feminine"]
        if marker in text
    ]

    if masculine and feminine:
        classification = "mixed_markers"

    elif masculine:
        classification = "masculine_markers_detected"

    elif feminine:
        classification = "feminine_markers_detected"

    else:
        classification = "no_selected_marker_detected"

    return {
        "classification": classification,
        "masculine_markers": masculine,
        "feminine_markers": feminine,
        "masculine_count": len(masculine),
        "feminine_count": len(feminine)
    }


def run_bias_evaluation(
    dataset,
    translation_function,
    target_language,
    model_name
):
    """
    Translate dataset sentences and screen outputs
    for selected grammatical gender markers.
    """

    if target_language not in ["Hindi", "Marathi"]:
        raise ValueError(
            "Target language must be Hindi or Marathi."
        )

    required_columns = [
        "id",
        "gender_condition",
        "profession",
        "english"
    ]

    missing = [
        column for column in required_columns
        if column not in dataset.columns
    ]

    if missing:
        raise ValueError(
            f"Missing dataset columns: {missing}"
        )

    results = []

    for _, row in dataset.iterrows():

        try:
            output = translation_function(
                row["english"],
                target_language
            )

            if isinstance(output, dict):
                success = output.get("success", False)
                translated = output.get("translated") or ""
                error = output.get("error") or ""

            else:
                success = bool(output)
                translated = str(output) if output else ""
                error = ""

            if success:
                marker_result = detect_gender_markers(
                    translated,
                    target_language
                )
                classification = marker_result["classification"]
                masculine = marker_result["masculine_markers"]
                feminine = marker_result["feminine_markers"]

            else:
                classification = "not_evaluated"
                masculine = []
                feminine = []

            results.append({
                "id": row["id"],
                "gender_condition": row["gender_condition"],
                "profession": row["profession"],
                "english": row["english"],
                "target_language": target_language,
                "model": model_name,
                "translation": translated,
                "success": success,
                "marker_classification": classification,
                "masculine_markers": ", ".join(masculine),
                "feminine_markers": ", ".join(feminine),
                "error": error,
                "human_review": "Pending"
            })

        except Exception as exc:

            results.append({
                "id": row["id"],
                "gender_condition": row["gender_condition"],
                "profession": row["profession"],
                "english": row["english"],
                "target_language": target_language,
                "model": model_name,
                "translation": "",
                "success": False,
                "marker_classification": "not_evaluated",
                "masculine_markers": "",
                "feminine_markers": "",
                "error": str(exc),
                "human_review": "Pending"
            })

    return pd.DataFrame(results)


def summarize_bias_results(results):
    """Summarize detected markers by gender condition."""

    successful = results[
        results["success"] == True
    ].copy()

    if successful.empty:
        return pd.DataFrame()

    return (
        successful.groupby(
            [
                "target_language",
                "model",
                "gender_condition",
                "marker_classification"
            ]
        )
        .size()
        .reset_index(name="count")
    )


def compare_neutral_professions(results):
    """
    Compare selected gender markers in translations
    of gender-neutral English inputs.
    """

    neutral = results[
        results["gender_condition"] == "neutral"
    ].copy()

    if neutral.empty:
        return pd.DataFrame()

    return (
        neutral.groupby(
            [
                "target_language",
                "model",
                "profession",
                "marker_classification"
            ]
        )
        .size()
        .reset_index(name="count")
    )
'''


# ==========================================
# 3. SAVE MODULE
# ==========================================

module_path = MODULE_DIR / "bias_evaluation.py"

module_path.write_text(
    bias_code.strip() + "\n",
    encoding="utf-8"
)

print("\nBias evaluation module created successfully!")
print("Saved at:", module_path)
print("File exists:", module_path.exists())
print("File size:", module_path.stat().st_size, "bytes")
```

Connect MarianMT and Qwen

```python

import sys
import torch
import time
from pathlib import Path

PROJECT_DIR = Path("/content/LinguaBridge-AI")

if str(PROJECT_DIR) not in sys.path:
    sys.path.insert(0, str(PROJECT_DIR))

# Import your EXISTING Qwen translator
from modules.llm_translator import translate_with_llm

MARIAN_NAMES = {
    "Hindi": "Helsinki-NLP/opus-mt-en-hi",
    "Marathi": "Helsinki-NLP/opus-mt-en-mr"
}

bias_marian_cache = {}

def bias_marian_translate(text, language):

    if language not in MARIAN_NAMES:
        return {"success": False, "error": "Unsupported language"}

    if not isinstance(text, str) or not text.strip():
        return {"success": False, "error": "Empty input"}

    try:
        # Reuse Module 1 models if they are already loaded
        if "loaded_models" in globals() and language in loaded_models:

            tokenizer = loaded_models[language]["tokenizer"]
            model = loaded_models[language]["model"]

            device = next(model.parameters()).device

        else:
            # Fallback: load the Marian model only once
            if language not in bias_marian_cache:

                from transformers import MarianMTModel, MarianTokenizer

                model_name = MARIAN_NAMES[language]

                tokenizer = MarianTokenizer.from_pretrained(model_name)
                model = MarianMTModel.from_pretrained(model_name)

                device = "cuda" if torch.cuda.is_available() else "cpu"

                model.to(device)
                model.eval()

                bias_marian_cache[language] = {
                    "tokenizer": tokenizer,
                    "model": model,
                    "device": device
                }

            bundle = bias_marian_cache[language]

            tokenizer = bundle["tokenizer"]
            model = bundle["model"]
            device = bundle["device"]

        inputs = tokenizer(
            text,
            return_tensors="pt",
            truncation=True,
            max_length=512
        )

        inputs = {
            key: value.to(device)
            for key, value in inputs.items()
        }

        start = time.time()

        with torch.inference_mode():
            output = model.generate(
                **inputs,
                max_new_tokens=160,
                num_beams=4
            )

        translated = tokenizer.decode(
            output[0],
            skip_special_tokens=True
        )

        return {
            "success": True,
            "translated": translated,
            "model": MARIAN_NAMES[language],
            "latency_seconds": round(time.time() - start, 3)
        }

    except Exception as exc:
        return {
            "success": False,
            "error": str(exc)
        }


def bias_qwen_translate(text, language):

    return translate_with_llm(
        text=text,
        target_language=language,
        style="Neutral",
        technique="zero_shot",
        max_new_tokens=160
    )


print("MarianMT adapter ready.")
print("Qwen adapter ready.")
print("GPU available:", torch.cuda.is_available())
```

Run the gender-bias experiment

```python

import pandas as pd

from modules.bias_evaluation import (
    run_bias_evaluation,
    summarize_bias_results,
    compare_neutral_professions
)

PROJECT_DIR = "/content/LinguaBridge-AI"

bias_df = pd.read_csv(
    f"{PROJECT_DIR}/data/bias_test_dataset.csv"
)

models = {
    "MarianMT": bias_marian_translate,
    "Qwen2.5-1.5B": bias_qwen_translate
}

all_bias_results = []

for language in ["Hindi", "Marathi"]:

    for model_name, translation_function in models.items():

        print("\n" + "=" * 60)
        print("LANGUAGE:", language)
        print("MODEL:", model_name)
        print("=" * 60)

        results = run_bias_evaluation(
            dataset=bias_df,
            translation_function=translation_function,
            target_language=language,
            model_name=model_name
        )

        all_bias_results.append(results)

        successful = int(results["success"].sum())
        print("Successful:", successful, "/", len(results))

        display(results[[
            "id",
            "gender_condition",
            "profession",
            "translation",
            "marker_classification"
        ]])

bias_results = pd.concat(
    all_bias_results,
    ignore_index=True
)

print("\nTOTAL EXPERIMENT RECORDS:", len(bias_results))
print("Successful translations:", int(bias_results["success"].sum()))
print("Failed translations:", int((~bias_results["success"]).sum()))
```
